# **Machine Learning Pipelines**

## _aka Workflow orchestration_


1. **Ingestion** – Download the data  
2. **Transforming Data** – Filtering, outlier removal, etc.  
3. **Feature Engineering** – Prepare data for ML  
4. **Hyperparameter Tuning** – Find best parameters  
5. **Train Final Model** – Use best params to train model  
6. **Model Registry** – Save final model to registry

## **Tools**
### General purpose
1. **Airflow**
2. **Prefect**
3. **Mage**

### For ML
1. **Kubeflow pipelines**
2. **MLFlow pipelines**
